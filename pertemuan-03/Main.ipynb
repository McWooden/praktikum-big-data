```python
import os
import sys
import getpass

# 1. Kunci JAVA_HOME ke OpenJDK 11 (khusus sub-shell sesi Jupyter ini)
os.environ["JAVA_HOME"] = "/usr/lib/jvm/java-11-openjdk"

# 2. Daftarkan path Hadoop ke environment sub-shell
HADOOP_HOME = os.path.expanduser("~/hadoop")
os.environ["HADOOP_HOME"] = HADOOP_HOME
os.environ["HADOOP_INSTALL"] = HADOOP_HOME
os.environ["HADOOP_MAPRED_HOME"] = HADOOP_HOME
os.environ["HADOOP_COMMON_HOME"] = HADOOP_HOME
os.environ["HADOOP_HDFS_HOME"] = HADOOP_HOME
os.environ["YARN_HOME"] = HADOOP_HOME
os.environ["HADOOP_COMMON_LIB_NATIVE_DIR"] = f"{HADOOP_HOME}/lib/native"
os.environ["HADOOP_OPTS"] = f"-Djava.library.path={HADOOP_HOME}/lib/native"

# 3. Sisipkan bin dan sbin ke PATH agar perintah 'hdfs' langsung dapat dipanggil dari mana saja
hadoop_paths = f"{HADOOP_HOME}/bin:{HADOOP_HOME}/sbin"
if hadoop_paths not in os.environ["PATH"]:
    os.environ["PATH"] = f"{hadoop_paths}:{os.environ['PATH']}"

USERNAME = getpass.getuser()
print("=== STATUS ENVIRONMENT SESI JUPYTER ===")
print(f"User Aktif        : {USERNAME}")
print(f"Direktori Proyek  : {os.getcwd()}")
print(f"Target JAVA_HOME  : {os.environ['JAVA_HOME']}")
print(f"Lokasi Hadoop     : {os.environ['HADOOP_HOME']}")
print("Status            : Lingkungan siap & terisolasi penuh!")
```
