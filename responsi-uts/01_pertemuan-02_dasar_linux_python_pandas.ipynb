## 2. Eksplorasi Data dengan Pandas (Pendekatan Single-Machine Konvensional)
Pandas adalah library manipulasi data terpopuler di Python. Mari kita buat dataset simulasi penjualan UMKM dan lakukan eksplorasi awal:

```python
import pandas as pd
import numpy as np

# Membuat DataFrame dari Dictionary data penjualan
data_umkm = {
    "kota": ["Magelang", "Yogyakarta", "Semarang", "Magelang", "Solo", "Purworejo", "Magelang", "Yogyakarta"],
    "kategori": ["Elektronik", "Fashion", "Makanan", "Fashion", "Elektronik", "Makanan", "Elektronik", "Fashion"],
    "unit_terjual": [120, 85, 230, 45, 90, 310, 60, 110],
    "harga_satuan": [50000, 120000, 15000, 110000, 75000, 12000, 50000, 125000]
}

df = pd.DataFrame(data_umkm)
print("5 Baris Pertama Data UMKM:")
df.head()
```

### 2.1 Memeriksa Struktur Data & Ringkasan Statistik
Fungsi `info()` menampilkan tipe data tiap kolom dan mendeteksi missing value, sedangkan `describe()` menampilkan ringkasan statistik (rata-rata, nilai min/max):

```python
# Cek tipe data dan missing value
print("--- INFO STRUKTUR DATA ---")
df.info()

print()
print("--- STATISTIK DESKRIPTIF ---")
df.describe()
```

## 3. Manipulasi Kolom dan Filtering Data
Kita dapat menambahkan kolom baru hasil kalkulasi matematika dan menyaring baris berdasarkan kondisi logika:

```python
# 1. Menambahkan kolom total pendapatan (unit x harga)
df["total_pendapatan"] = df["unit_terjual"] * df["harga_satuan"]

# 2. Filtering kondisi tunggal: hanya kota Magelang
df_magelang = df[df["kota"] == "Magelang"]
print(f"Data Magelang ({len(df_magelang)} baris):")
display(df_magelang)

# 3. Filtering multi-kondisi: unit > 100 DAN kategori Elektronik
# Catatan: Wajib gunakan tanda kurung () untuk tiap kondisi!
df_khusus = df[(df["unit_terjual"] > 100) & (df["kategori"] == "Elektronik")]
print()
print("Data Khusus (Unit > 100 & Elektronik):")
display(df_khusus)
```

## 4. Agregasi Data (GroupBy)
Mengelompokkan data berdasarkan kota untuk melihat total omset dan rata-rata penjualan:

```python
# GroupBy berdasarkan kota
ringkasan_kota = df.groupby("kota").agg({
    "total_pendapatan": "sum",
    "unit_terjual": "mean"
}).reset_index().rename(columns={
    "total_pendapatan": "total_omset",
    "unit_terjual": "rata_rata_unit"
}).sort_values("total_omset", ascending=False)

print("Ringkasan Kinerja Penjualan per Kota:")
display(ringkasan_kota)
```

## ⚠️ 5. Titik Batas: Mengapa Pandas Tidak Cukup untuk Big Data?

Jika Pandas sangat nyaman digunakan, **kenapa kita butuh Hadoop dan Apache Spark?**

1. **Limitasi Fisik RAM (In-Memory Crash):**  
   Pandas memuat seluruh dataset langsung ke dalam memori RAM komputer sekaligus. Jika kamu punya file log sebesar 50 GB di laptop dengan RAM 8 GB, laptopmu akan langsung mengalami **`MemoryError`** atau freeze!
2. **Prinsip Single-Machine:**  
   Pandas hanya berjalan di 1 mesin (laptopmu saja). Jika data bertambah besar, satu-satunya cara adalah *Vertical Scaling* (beli RAM lebih mahal).
3. **Ketiadaan Pemrosesan Terdistribusi:**  
   Pandas tidak bisa otomatis memecah perhitungan ke 10 atau 100 komputer sekaligus.

> **Jembatan ke Modul 02:**  
> Untuk mengatasi batasan RAM ini, kita memerlukan **Apache Hadoop (HDFS)** sebagai sistem penyimpanan data terdistribusi di banyak komputer, yang akan kita pelajari dan uji coba di Modul 02!
